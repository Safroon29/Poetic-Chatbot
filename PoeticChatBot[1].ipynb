# Prompt Engineering

### Prompt Engineering

The system prompt defines the chatbot's personality and response
format.

It instructs the LLM to:
- Respond only in poetry
- Understand the user's emotion
- Use metaphors and imagery
- Maintain a specific length
- Avoid normal conversational responses
- Produce personalized poetic responses

This demonstrates prompt engineering by controlling the
LLM's behavior through carefully designed instructions.

```python
SYSTEM_PROMPT = """
You are a Poetic Personality Chatbot.

Your task is to transform every user's message into
beautiful, meaningful poetry.

Follow these rules:

1. Respond ONLY in poetry.
2. Generate 4 to 8 lines.
3. Understand and reflect the user's emotion.
4. Use metaphors, imagery and expressive language.
5. Use rhyme whenever it feels natural.
6. Do not explain your response.
7. Do not use bullet points.
8. Do not give a normal conversational answer.
9. Make the poem feel personal to the user.
10. End with a hopeful or thoughtful feeling whenever appropriate.
"""
```

### Poetic Response Function

The `generate_poem()` function sends the user's message
and the system prompt to the LLM.

The model generates a poetic response based on the
instructions provided in the system prompt.

The generated poem is then returned to the chatbot.

```python
def generate_poem(user_message):

    response = client.chat.completions.create(

        model="llama-3.3-70b-versatile",

        messages=[
            {
                "role": "system",
                "content": SYSTEM_PROMPT
            },
            {
                "role": "user",
                "content": user_message
            }
        ],

        temperature=0.9,
        max_tokens=200
    )

    return response.choices[0].message.content
```

# Test Chat

```python
user_message = "I am feeling happy today."

poem = generate_poem(user_message)

print(poem)
```

```python
print("=" * 55)
print("🌸 POETIC PERSONALITY CHATBOT 🌸")
print("=" * 55)
print("Type 'exit' to end the conversation.")
print("=" * 55)
conversation_history = []

while True:

    user = input("\nYou: ")

    if user.lower() == "exit":
          print("\n🌸 Until we meet again, may your days be filled with verses and your heart with light. 🌸")
          break


    poem = generate_poem(user)

    conversation_history.append({
        "user": user,
        "poem": poem
    })

    print("\nPoet:")
    print(poem)
```

# The History Page

```python
print("\n🌸 CONVERSATION HISTORY 🌸")
print("=" * 55)

for i, conversation in enumerate(conversation_history, 1):

    print(f"\nConversation {i}")
    print("You:", conversation["user"])
    print("Poet:")
    print(conversation["poem"])
```
